# Escalado
Conjunto de datos: iris (incluido en sklearn)
Este conjunto de datos es compacto y perfecto para demostrar técnicas de escalado.

```python
# Importar librerías necesarias
from sklearn.datasets import load_iris
from sklearn.preprocessing import MinMaxScaler, StandardScaler
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Cargar el conjunto de datos
iris = load_iris(as_frame=True)
data = iris.data
```

```python
# Visualizar las estadísticas iniciales
print("Estadísticas iniciales:")
data.describe()
```

```python
# Aplicar MinMaxScaler
minmax_scaler = MinMaxScaler()
data_minmax = pd.DataFrame(minmax_scaler.fit_transform(data), columns=data.columns)
```

```python
# Aplicar StandardScaler
standard_scaler = StandardScaler()
data_standard = pd.DataFrame(standard_scaler.fit_transform(data), columns=data.columns)
```

### 1. ¿Qué diferencias encuentras entre las estadísticas originales y las del escalado MinMax?


```python
# Comparación de estadísticas
print("--- Estadísticas Originales ---")
print(data.describe().loc[['min', 'max', 'mean', 'std']])

print("\n--- Estadísticas MinMax (Escalado) ---")
print(data_minmax.describe().loc[['min', 'max', 'mean', 'std']])
```

### 2. Crea una gráfica de la distribución de la variable 'sepal length (cm)' antes y después del escalado.

```python
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Antes del escalado
ax1.hist(data['sepal length (cm)'], bins=20, color='skyblue', edgecolor='black')
ax1.set_title('Original: Sepal Length')
ax1.set_xlabel('Centímetros')

# Después del escalado (MinMax)
ax2.hist(data_minmax['sepal length (cm)'], bins=20, color='salmon', edgecolor='black')
ax2.set_title('MinMax Scaled: Sepal Length')
ax2.set_xlabel('Valor Escala [0,1]')

plt.tight_layout()
plt.show()
```

### 3. Crea una gráfica de dispersión entre 'sepal length (cm)' y 'sepal width (cm)' con los datos escalados por StandardScaler.

```python
plt.figure(figsize=(8, 6))
plt.scatter(data_standard['sepal length (cm)'], 
            data_standard['sepal width (cm)'], 
            c=iris.target, cmap='viridis', alpha=0.7)

plt.axhline(0, color='red', linestyle='--', alpha=0.3) # Eje central Y
plt.axvline(0, color='red', linestyle='--', alpha=0.3) # Eje central X

plt.title('Dispersión Sepal (StandardScaler)')
plt.xlabel('Sepal Length (Z-score)')
plt.ylabel('Sepal Width (Z-score)')
plt.grid(True, alpha=0.3)
plt.show()
```
